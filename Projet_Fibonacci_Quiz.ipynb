## 1. La suite de Fibonacci

La suite est définie par :

- F(0) = 0
- F(1) = 1
- F(n) = F(n-1) + F(n-2)

Exemples : 0, 1, 1, 2, 3, 5, 8, 13, ...

```python
for n in range(0, 11):
    print(f"n={n} -> {nth_fibonacci_memoized(n)}")
```

## 2. Comparaison des approches

### Version naïve
- récursive
- très lente
- complexité : O(2^n)

### Version mémoïsée
- récursive avec cache
- évite les recalculs
- complexité : O(n)

### Version itérative
- bottom-up
- utilise un tableau ou seulement deux variables
- performant et sans risque de dépassement de pile

```python
import time

def benchmark(func, n):
    start = time.perf_counter()
    result = func(n)
    elapsed = time.perf_counter() - start
    return result, elapsed

n = 40
for name, func in [
    ('naive', nth_fibonacci),
    ('memoized', nth_fibonacci_memoized),
    ('iterative_table', fib),
    ('iterative_opt', nthFibonacci),
]:
    result, elapsed = benchmark(func, n)
    print(f"{name:15} -> valeur={result} | temps={elapsed:.6f}s")
```

## 3. À retenir

- La version naïve est simple mais exponentielle.
- La mémoïsation réduit fortement les recalculs.
- Le bottom-up est souvent la meilleure solution en Python pour un grand n.
- La récursion peut provoquer `RecursionError` si n est trop grand.
- Les arguments mutables en valeur par défaut doivent être évités dans les fonctions de cache.

## 4. Conclusion

Ce projet montre clairement qu'une solution naïve peut sembler élégante, mais qu'il faut toujours penser à la complexité algorithmique et aux limites de la pile d'appels. La mémoïsation et le bottom-up sont des outils essentiels pour écrire des solutions efficaces et robustes.
